```python
import ast

def parse_line(line):
    sentence, triplet_str = line.strip().split("####")
    triplets = ast.literal_eval(triplet_str)  # safely parses the string as a Python list
    return sentence, triplets

# parse the whole file
parsed = [parse_line(l) for l in lines]

# Derive UABSA view: (aspect, sentiment) — drop category
uabsa_data = [
    (sent, [(asp, sent_lbl) for asp, cat, sent_lbl in triplets])
    for sent, triplets in parsed
]

# Derive TASD view: (aspect, category, sentiment) — keep as-is
tasd_data = parsed  # already in this format

# Count NULL aspect cases
null_count = 0
total_triplets = 0
for sent, triplets in parsed:
    for asp, cat, sent_lbl in triplets:
        total_triplets += 1
        if asp == 'NULL':
            null_count += 1

print(f"{null_count} / {total_triplets} triplets have NULL aspect ({100*null_count/total_triplets:.2f}%)")
```

```python
def null_rate(filepath):
    path = hf_hub_download(repo_id="Multilingual-NLP/M-ABSA", filename=filepath, repo_type="dataset")
    with open(path, encoding="utf-8") as f:
        lines = f.readlines()
    parsed = [parse_line(l) for l in lines]
    null_count = sum(1 for _, triplets in parsed for asp, cat, sent_lbl in triplets if asp == 'NULL')
    total = sum(len(triplets) for _, triplets in parsed)
    return null_count, total

for split in ["train", "dev", "test"]:
    n, t = null_rate(f"restaurant/en/{split}.txt")
    print(f"{split}: {n}/{t} ({100*n/t:.2f}%)")
```

```python
import pandas as pd
import ast
from huggingface_hub import list_repo_files, hf_hub_download

def parse_line(line):
    sentence, triplet_str = line.strip().split("####")
    triplets = ast.literal_eval(triplet_str)
    return sentence, triplets

all_files = list_repo_files("Multilingual-NLP/M-ABSA", repo_type="dataset")
data_files = [f for f in all_files if f.endswith(".txt")]

print(f"Found {len(data_files)} data files")

rows = []
skipped_lines = []
skipped_triplets = []

for filepath in data_files:
    parts = filepath.split("/")
    if len(parts) != 3:
        continue
    domain, language, split_file = parts
    split = split_file.replace(".txt", "")

    local_path = hf_hub_download(
        repo_id="Multilingual-NLP/M-ABSA",
        filename=filepath,
        repo_type="dataset"
    )

    with open(local_path, encoding="utf-8") as f:
        lines = f.readlines()

    for line_num, line in enumerate(lines):
        line = line.strip()
        if not line:
            continue
        try:
            sentence, triplets = parse_line(line)
        except (ValueError, SyntaxError) as e:
            skipped_lines.append({
                "file": filepath,
                "line_num": line_num,
                "raw_line": line,
                "error": str(e)
            })
            continue

        for triplet in triplets:
            if len(triplet) != 3:
                skipped_triplets.append({
                    "file": filepath,
                    "line_num": line_num,
                    "triplet": triplet,
                    "triplet_len": len(triplet)
                })
                continue
            aspect, category, sentiment = triplet
            rows.append({
                "domain": domain,
                "language": language,
                "split": split,
                "sentence": sentence,
                "aspect": aspect,
                "category": category,
                "sentiment": sentiment,
                "is_null_aspect": aspect == "NULL"
            })

print(f"Total rows parsed: {len(rows)}")
print(f"Total lines skipped (parse errors): {len(skipped_lines)}")
print(f"Total triplets skipped (wrong length): {len(skipped_triplets)}")
```
